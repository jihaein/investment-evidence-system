```python
# SPIKE Test (Open Dart)
import os
from dotenv import load_dotenv
import requests
from datetime import datetime, timedelta

load_dotenv()
KEY = os.getenv("OPENDART_API_KEY")

today = datetime.now().strftime("%Y%m%d")
start = (datetime.now() - timedelta(days=120)).strftime("%Y%m%d")

params = {
    "crtfc_key": KEY,
    "corp_code": "00174208",   # 진성티이씨 (네 데이터의 corp_code)
    "bgn_de": start,
    "end_de": today,
    "pblntf_ty": "A",          # 정기공시 = 사업·반기·분기보고서
    "page_count": 100,
}
r = requests.get("https://opendart.fss.or.kr/api/list.json", params=params)
data = r.json()

print("status:", data.get("status"), data.get("message"))
for item in data.get("list", []):
    print(item["rcept_dt"], item["report_nm"], item["rcept_no"])
```
